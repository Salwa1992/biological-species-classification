```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, f1_score, recall_score, roc_auc_score

iris_data = load_iris(as_frame=True)
df = iris_data.frame


df['Is_Virginica'] = (iris_data.target == 2).astype(int)

X = df.drop(columns=['target', 'Is_Virginica'])
y = df['Is_Virginica']

print(f"Rows: {X.shape[0]}, Features: {X.shape[1]}")

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```
